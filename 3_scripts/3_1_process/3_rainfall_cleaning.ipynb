```r
# Project: Ottawa Stormwater Capstone
# Purpose: Prepare historical rainfall data

# 1. Load the package
library(readr)
library(dplyr)
```

```r
# 2. Import the original dataset
rainfall_raw <- read_csv("/content/1_rainfall_raw.csv")
```

```r
# 3. Print the dataset

rainfall_raw
```

```r
# 3. Add row number as the first column
rainfall_raw_rows <- rainfall_raw %>%
  mutate(Row = row_number()) %>%
  relocate(Row, .before = 1)

rainfall_raw_rows
```

```r
# 3. Examine the data
summary(rainfall_raw)
```
